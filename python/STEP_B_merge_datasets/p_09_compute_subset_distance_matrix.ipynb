## This script is used to compute a subset of the in “compute_distance_matrix.ipynb” produced distance matrix "distmatrix.pkl" 
## subset is computed by implementing a threshold of maximum distance in km
## adapt thrs as it make sense for analysis of duplicated stations--> suggestions 0.25km or 0.5km


```python
# load required modules:

import pandas as pd
from glob import glob
import numpy as np
import geopandas as gpd
import matplotlib.pyplot as plt
from shapely.geometry import Point
import warnings
from tqdm import trange
from time import time
import pickle
from scipy.spatial.distance import cdist
from scipy.spatial import distance_matrix
import pdb
from math import sin, cos, sqrt, atan2, radians
from scipy.spatial.distance import pdist, squareform
from sklearn.metrics.pairwise import pairwise_distances
import multiprocessing
from multiprocessing import Pool
n_cores = multiprocessing.cpu_count()
import os
from tqdm import tqdm
from time import time
```

```python
# ---load distance matrix ----
tic = time()
dist_matrix = pickle.load(open('../../output_data/distmatrix.pkl', 'rb'))
print("Elapsed %.2f seconds" % ((time() - tic)))
#print(dist_matrix.shape,stations_stats.shape)     
```

```python
# find all stations which are in vicinity of other station--> distance of <= 0.25km 

thrs = 0.25 # or 0.5km
# select all columns of stations by adding True, if any distance is smaller threshold of 0.5km/0.25km
stations_duplicatedkm_bool = ((dist_matrix <= thrs).sum(axis=1)>1)
stations_dupkm = dist_matrix[stations_duplicatedkm_bool]
# and select only columns of duplicated stations
stations_dupkm_2 = stations_dupkm.loc[:,stations_duplicatedkm_bool]

```

```python
# SAVE SUBSET OF DISTANCE MATRIX:
# define the output folder path:
output_folder = '../../output_data/merged_datasets'

os.chdir(f'{output_folder}')
tic = time()
pickle.dump(stations_dupkm_2,open(f'/duplicated_only_distmatrix_{thrs}km.pkl','wb'))
print("Elapsed %.2f seconds" % ((time() - tic)))
```
