## This script merges raw data of different datasets to one main file:



```python
import pandas as pd
from glob import glob
import numpy as np
import geopandas as gpd
import matplotlib.pyplot as plt
from shapely.geometry import Point
import os
```

```python
# define the output folder path:

output_folder = '../../output_data/merged_datasets'

# Check if the folder exists, if not, create it:
if not os.path.exists(output_folder):
    os.makedirs(output_folder)
```

```python
#--- List files ---
inputFolder = '../../output_data/outlier_removed_raw_data/outlier_removed/'
files = glob(f'{inputFolder}*csv')
print([file.split('_')[-1].split('.')[0] for file in files])
```

```python
# --- Merge data ---
list_out = []
for file in files:
    dataset_name = file.split('_')[-1].split('.')[0]
    delimiter = ','
    decimal = '.'
    raw_data = pd.read_csv(file,delimiter=delimiter,decimal=decimal)


    to_replace = {'SiteID':'site_id','Station':'site_id','date_string':'obs_date','Date':'obs_date'}
    raw_data = raw_data.rename(columns=to_replace)
    
    #--- Print some information ---
    print('\n',dataset_name,raw_data.columns.tolist())
    
    #--- Append datasets ---
    raw_data['dataset'] = dataset_name
    raw_data = raw_data[raw_data.columns.drop('Unnamed: 0')]
    list_out.append(raw_data.copy())

```

#### save merged data frame to csv file:

```python
merged_raw_data = pd.concat(list_out,axis=0)
merged_raw_data = merged_raw_data.replace(np.nan,-9999)
# replace '#VALUE!' cells by -9999 ---- 
(merged_raw_data[4:]=='#VALUE!').sum()
merged_raw_data = merged_raw_data.replace('#VALUE!',-9999)

merged_raw_data.to_csv(f'{output_folder}/merged_raw_data_outliers_removed.csv')
```
