
### This script assigns water quality stations to the nearest HydroRiver segments.

#### 1) In the first block, the closest and second-closest segments are identified for each station
####    using the ArcPy function: arcpy.analysis.GenerateNearTable.

#### 2) In the second block, the closest segment is assigned to each station using arcpy.analysis.Near.
####    The result is saved as a new shapefile.

#### Note: If the script fails to run, ensure that ArcMap and ArcPy are properly installed.

```python
# load arcpy module: 
import arcpy  
import os  
```

```python
# define the output folder path:

output_folder = '../../output_data/assign_stations_HydroRivers'

# Check if the folder exists, if not, create it:
if not os.path.exists(output_folder):
    os.makedirs(output_folder)
```

```python
# shapefiles
input_shapefile_stations = "../../output_data/merged_datasets/stations.shp"
near_features_river_segments = "../../input_data/geodata/HydroRIVERS/HydroRIVERS_v10_shp/HydroRIVERS_v10.shp"
out_table = "../../output_data/assign_stations_HydroRivers/stations_assigned_rivers_geodesic.csv"
```

## 1. Following first block generates a table containing closest and second closest river segment (HydroRivers) for each station:

```python
# 
 
 # arcpy.GenerateNearTable_analysis to get also second nearest stream segment

arcpy.analysis.GenerateNearTable(in_features = input_shapefile_stations, near_features = near_features_river_segments,out_table = out_table, location = 'NO_LOCATION', angle = 'NO_ANGLE', closest = 'ALL', closest_count = 2, method = 'GEODESIC')

print("CSV file saved to: {}.".format(out_table))
```


## 2. In this step, the stations shapefile is copied and processed using arcpy.analysis.near().
## The tool appends the ID and distance of the nearest HydroRiver segment to each station 
## in the attribute table.

```python
# 2. assign closest segment to each station arcpy.analysis.Near

if not arcpy.Exists("../../output_data/assign_stations_HydroRivers/ArcPy_output"):
    arcpy.management.CreateFolder("../../output_data/assign_stations_HydroRivers", 'ArcPy_output')

# set workspace
arcpy.env.workspace = "../../output_data/assign_stations_HydroRivers/ArcPy_output"



if not arcpy.Exists('stations_assigned_river.shp'):
    arcpy.management.Copy(input_shapefile_stations, 'stations_assigned_river')
```

```python

in_shp2 ="../../output_data/assign_stations_HydroRivers/ArcPy_output/stations_assigned_river.shp"
arcpy.analysis.Near(in_features = in_shp2, near_features = near_features_river_segments, method ='GEODESIC')
```
