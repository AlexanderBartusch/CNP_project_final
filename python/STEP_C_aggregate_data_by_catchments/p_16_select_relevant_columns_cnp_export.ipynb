### Description: 
#### This script filters export_data_daily to make file size smaller.
#### The csv-files containing exported nutrient data are quite large, this script reduces number of columns. 
#### only median columns in mg/L, obs_date and HYBAS_ID  are selected.


```python
import pandas as pd
import numpy as np
```

```python
# import data:
exp_data_daily_SRP = pd.read_csv("../../output_data/CNP_data_catchments/DOC_TOC_DIN_SRP_daily_median_sd_subbasins.csv")
exp_data_daily_TP  = pd.read_csv("../../output_data/CNP_data_catchments/DOC_TOC_DIN_TP_daily_median_sd_subbasins.csv")
```

```python
columns_to_select = [
    "HYBAS_ID", "obs_date", "NO3N_median", "NO2N_median", "NH4N_median",
    "NO2N_NO3N_median", "TOC_median", "DOC_median", "DOC_TOC_median",
    "DIN_median", "SRP_median"
]

exp_data_daily_SRP_selection = exp_data_daily_SRP[columns_to_select]

# For TP, replace SRP_median with TP_median
columns_to_select_TP = columns_to_select[:-1] + ["TP_median"]

exp_data_daily_TP_selection = exp_data_daily_TP[columns_to_select_TP]
```

#### Save data:

```python
exp_data_daily_SRP_selection.to_csv(
    "../../output_data/CNP_data_catchments/basin_exp_DOC_TOC_DIN_SRP_daily_median_mgL_selected.csv",
    index=False
)

exp_data_daily_TP_selection.to_csv(
    "../../output_data/CNP_data_catchments/basin_exp_DOC_TOC_DIN_TP_daily_median_mgL_selected.csv",
    index=False
)
```
