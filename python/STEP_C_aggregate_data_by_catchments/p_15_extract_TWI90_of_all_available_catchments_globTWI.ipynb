```python
import pandas as pd
from rasterstats import zonal_stats
import geopandas as gpd
import os
```

```python
# define the output folder path:

output_folder = '../../output_data/TWI_QGIS_and_TWI90_extraction'

# Check if the folder exists, if not, create it:
if not os.path.exists(output_folder):
    os.makedirs(output_folder)
```

```python
twi90_basins_SRP = zonal_stats("../../output_data/CNP_data_catchments/DOC_TOC_DIN_SRP_daily_median_sd_subbasins.shp", "../../input_data/geodata/TWI_global.tif", stats=['percentile_90'])
```

```python
df_twi90 = pd.DataFrame(twi90_basins_SRP).rename(columns = {'percentile_90':'twi90'})
```

```python
basins = gpd.read_file("../../output_data/CNP_data_catchments/DOC_TOC_DIN_SRP_daily_median_sd_subbasins.shp")
```

```python
basin_ids = basins['HYBAS_ID']
```
